**1)Print customer Id, customer name and average account_balance maintained by each customer for all  of his/her accounts in the bank.(8 Rows)**


```python
query = """
SELECT
    bc.customer_id,
    bc.customer_name,
    AVG(ba.Balance_amount) AS average_balance
FROM BANK_CUSTOM1R bc, BANK_ACCOUNT ba
WHERE bc.customer_id = ba.Customer_id
GROUP BY bc.customer_id, bc.customer_name;
"""
result = pd.read_sql_query(query, conn)
print(tabulate(result, headers='keys', tablefmt='psql'))
```

**2) Print customer_id , account_number and balance_amount , condition that if balance_amount is nil then assign transaction_amount  for account_type = "Credit Card"(4 Rows).**

```python
query = """
SELECT
    a.customer_id,
    a.account_number,
    CASE
        WHEN a.balance_amount = 0 THEN t.transaction_amount
        ELSE a.balance_amount
    END AS balance_amount
FROM Bank_Account_Details a
JOIN (
    SELECT account_number, transaction_amount, transaction_date
    FROM BANK_ACCOUNT_TRANSACTION
    WHERE (account_number, transaction_date) IN (
        SELECT account_number, MAX(transaction_date)
        FROM BANK_ACCOUNT_TRANSACTION
        GROUP BY account_number
    )
    GROUP BY account_number, transaction_amount, transaction_date
) t
    ON a.account_number = t.account_number
WHERE a.account_type = 'Credit Card';
"""

result = pd.read_sql_query(query, conn)

print(tabulate(result, headers='keys', tablefmt='psql'))
```

**3) Print account_number and balance_amount , transaction_amount,Transaction_Date from Bank_Account_Details and bank_account_transaction for all the transactions occurred during march,2020 and april, 2020(12 Rows)**

```python
query = """
SELECT
    a.account_number,
    a.balance_amount,
    t.transaction_amount,
    t.transaction_date
FROM Bank_Account_Details a
JOIN BANK_ACCOUNT_TRANSACTION t
    ON a.account_number = t.account_number
WHERE t.transaction_date >= '2020-03-01'
  AND t.transaction_date < '2020-05-01'
GROUP BY
    a.account_number,
    a.balance_amount,
    t.transaction_amount,
    t.transaction_date
ORDER BY t.transaction_date;
"""

result = pd.read_sql_query(query, conn)

print(tabulate(result, headers='keys', tablefmt='psql'))
```

**4) Print all of the customer id, account number,  balance_amount, transaction_amount , Transaction_Date  from bank_customer, Bank_Account_Details and bank_account_transaction tables where excluding all of their transactions in march, 2020  month (22 Rows)**

```python
query = """
SELECT DISTINCT
    c.customer_id,
    a.account_number,
    a.balance_amount,
    t.transaction_amount,
    t.transaction_date
FROM BANK_CUSTOM1R c
JOIN (
    SELECT DISTINCT
        customer_id,
        account_number,
        balance_amount
    FROM Bank_Account_Details
) a
    ON c.customer_id = a.customer_id
JOIN BANK_ACCOUNT_TRANSACTION t
    ON a.account_number = t.account_number
WHERE t.transaction_date < '2020-03-01'
   OR t.transaction_date >= '2020-04-01'
ORDER BY t.transaction_date;
"""

result = pd.read_sql_query(query, conn)
print(tabulate(result, headers='keys', tablefmt='psql'))
```

**5) Print only the customer id, account_number, balance_amount,transaction_amount ,transaction_date who did transactions during the first quarter. Do not display the accounts if they have not done any transactions in the first quarter.(16 Rows)**

```python
query = """
SELECT DISTINCT
    c.customer_id,
    a.account_number,
    a.balance_amount,
    t.transaction_amount,
    t.transaction_date
FROM BANK_CUSTOM1R c
JOIN (
    SELECT DISTINCT
        customer_id,
        account_number,
        balance_amount
    FROM Bank_Account_Details
) a
    ON c.customer_id = a.customer_id
JOIN BANK_ACCOUNT_TRANSACTION t
    ON a.account_number = t.account_number
WHERE t.transaction_date >= '2020-01-01'
  AND t.transaction_date < '2020-04-01'
ORDER BY t.transaction_date;
"""

result = pd.read_sql_query(query, conn)
print(tabulate(result, headers='keys', tablefmt='psql'))
```

**6) Print account_number, Event and Customer_message from BANK_CUSTOMER_MESSAGES and Bank_Account_Details to display an “Adhoc" Event for all customers who have  “SAVINGS" account_type account.(8 Rows)**

```python
query = """
SELECT DISTINCT
    a.account_number,
    m.Event,
    m.Customer_message
FROM Bank_Account_Details a
CROSS JOIN BANK_CUSTOMER_MESSAGES m
WHERE a.account_type = 'SAVINGS'
  AND m.Event = 'Adhoc'
ORDER BY a.account_number;
"""

result = pd.read_sql_query(query, conn)
print(tabulate(result, headers='keys', tablefmt='psql'))
```

**7) Print all Customer_id, Account_Number, Account_type, and display deducted balance_amount by  subtracting only negative transaction_amounts for Relationship_type ="P" ( P - means  Primary , S - means Secondary ) .(27 Rows)**

```python
query = """
SELECT
    a.Customer_id,
    a.Account_Number,
    a.Account_type,
    a.Balance_amount + t.transaction_amount AS deducted_balance_amount
FROM BANK_ACCOUNT a
JOIN (
    SELECT DISTINCT
        account_number,
        transaction_amount,
        transaction_date
    FROM BANK_ACCOUNT_TRANSACTION
    WHERE transaction_amount < 0
) t
    ON a.Account_Number = t.account_number
WHERE a.Relation_ship = 'P'
ORDER BY a.Customer_id, a.Account_Number, t.transaction_date;
"""

result = pd.read_sql_query(query, conn)
print(tabulate(result, headers='keys', tablefmt='psql'))
```

```python
#8)b)
query = """
SELECT DISTINCT
    a.Account_Number,
    a.Account_type,
    t.transaction_amount,
    r.Linking_Account_Number,
    l.Account_type AS Linking_Account_Type
FROM BANK_ACCOUNT a
JOIN BANK_ACCOUNT_TRANSACTION t
    ON a.Account_Number = t.account_number
JOIN Bank_Account_Relationship r
    ON a.Account_Number = r.Account_Number
LEFT JOIN BANK_ACCOUNT l
    ON r.Linking_Account_Number = l.Account_Number
ORDER BY a.Account_Number, t.transaction_amount;
"""

result = pd.read_sql_query(query, conn)
print(tabulate(result, headers='keys', tablefmt='psql'))
```

```python
#9)c)
query = """
SELECT DISTINCT
    r.Account_Number,
    r.Account_type,
    r.Linking_Account_Number,
    l.Account_type AS Linking_Account_Type,
    t1.transaction_amount AS Account_Transaction_Amount,
    t2.transaction_amount AS Linking_Account_Transaction_Amount
FROM Bank_Account_Relationship r

LEFT JOIN BANK_ACCOUNT_TRANSACTION t1
    ON r.Account_Number = t1.account_number

LEFT JOIN BANK_ACCOUNT l
    ON r.Linking_Account_Number = l.Account_Number

LEFT JOIN BANK_ACCOUNT_TRANSACTION t2
    ON r.Linking_Account_Number = t2.account_number

WHERE t1.transaction_amount IS NOT NULL
   OR t2.transaction_amount IS NOT NULL

ORDER BY r.Account_Number, r.Linking_Account_Number;
"""

result = pd.read_sql_query(query, conn)
print(tabulate(result, headers='keys', tablefmt='psql'))
```

**10) Display all saving account holders have “Add-on Credit Cards" and “Credit cards" (3 Rows)**

```python
query = """
SELECT DISTINCT
    s.Customer_id,
    s.Account_Number AS Savings_Account_Number,
    s.Account_type AS Savings_Account_Type,
    c.Account_Number AS Credit_Card_Account_Number,
    c.Account_type AS Credit_Card_Type,
    ad.Account_Number AS Addon_Card_Account_Number,
    ad.Account_type AS Addon_Card_Type
FROM BANK_ACCOUNT s
JOIN Bank_Account_Relationship cr
    ON cr.Linking_Account_Number = s.Account_Number
   AND cr.Account_type = 'Credit Card'
JOIN Bank_Account_Relationship ar
    ON ar.Linking_Account_Number = cr.Account_Number
   AND ar.Account_type = 'Add-on Credit Card'
LEFT JOIN BANK_ACCOUNT c
    ON c.Account_Number = cr.Account_Number
LEFT JOIN BANK_ACCOUNT ad
    ON ad.Account_Number = ar.Account_Number
WHERE s.Account_type = 'SAVINGS'
ORDER BY s.Customer_id;
"""

result = pd.read_sql_query(query, conn)
print(tabulate(result, headers='keys', tablefmt='psql'))
```

```python
query = """
SELECT
    SUM(t.transaction_amount) AS Total_Credit_Card_Transaction_Amount
FROM (
    SELECT DISTINCT
        Account_Number,
        Account_type,
        Linking_Account_Number
    FROM Bank_Account_Relationship
    WHERE Account_type = 'Credit Card'
) r
JOIN (
    SELECT DISTINCT
        account_number,
        transaction_amount,
        transaction_date
    FROM BANK_ACCOUNT_TRANSACTION
) t
    ON r.Account_Number = t.account_number
JOIN (
    SELECT DISTINCT
        Account_Number,
        Account_type
    FROM BANK_ACCOUNT
    WHERE Account_type = 'SAVINGS'
) s
    ON r.Linking_Account_Number = s.Account_Number;
"""

result = pd.read_sql_query(query, conn)
print(tabulate(result, headers='keys', tablefmt='psql'))
```
